# LDA - Gradient Boosting

```python
import numpy as np
from sklearn.metrics import classification_report, confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

# --- LDA PIPELINE FOR estilo_merged ---
if 'estilo_merged' not in df_recommender.columns:
    raise ValueError("A coluna 'estilo_merged' não foi encontrada em df_recommender. Verifique a etapa de mapeamento de estilos.")

target_column = 'estilo_merged_v2'
X_style = df_recommender.drop(columns=[target_column])
y_style = df_recommender[target_column].astype(str).str.strip().replace({'': 'missing', 'nan': 'missing'})
y_style = y_style.fillna('missing').astype('category')

print(f"Usando coluna alvo para Estilo: {target_column}")
print(f"Número de classes: {y_style.nunique()}")
print(y_style.value_counts())

# Excluir identificadores do conjunto de features se não for útil
if 'ID_ARTISTA' in X_style.columns:
    X_style = X_style.drop(columns=['ID_ARTISTA'])

numerical_cols = X_style.select_dtypes(include=['number']).columns.tolist()
categorical_cols = X_style.select_dtypes(include=['object', 'category']).columns.tolist()

print(f"Colunas numéricas: {len(numerical_cols)}")
print(f"Colunas categóricas: {len(categorical_cols)}")

preprocessor_style = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), numerical_cols),
        ('cat', OneHotEncoder(handle_unknown='ignore', sparse_output=False), categorical_cols)
    ],
    remainder='passthrough'
)

lda_pipeline_style = Pipeline(steps=[
    ('preprocessor', preprocessor_style),
    ('lda', LinearDiscriminantAnalysis(solver='svd'))
])

X_train_style, X_test_style, y_train_style, y_test_style = train_test_split(
    X_style, y_style, test_size=0.2, random_state=42, stratify=y_style
)

print("Treinando o modelo LDA para estilo_merged...")
lda_pipeline_style.fit(X_train_style, y_train_style)
print("Modelo LDA para estilo_merged treinado com sucesso!")

X_lda_style = lda_pipeline_style.transform(X_style)
print(f"Dimensão original: {X_style.shape}")
print(f"Dimensão após LDA para estilo_merged: {X_lda_style.shape}")

# Avaliação do classificador LDA para estilo_merged
y_pred_style = lda_pipeline_style.predict(X_test_style)
print("\nRelatório de classificação para estilo_merged:")
print(classification_report(y_test_style, y_pred_style, zero_division=0))

cm_style = confusion_matrix(y_test_style, y_pred_style, labels=lda_pipeline_style.named_steps['lda'].classes_)
fig, ax = plt.subplots(figsize=(10, 8))
ConfusionMatrixDisplay(confusion_matrix=cm_style, display_labels=lda_pipeline_style.named_steps['lda'].classes_).plot(ax=ax, cmap='Blues', colorbar=False)
ax.set_title('Confusion Matrix - LDA estilo_merged')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()

# Salvar componentes LDA para estilo_merged
feature_names_style = preprocessor_style.get_feature_names_out()
df_lda_components_style = pd.DataFrame(
    lda_pipeline_style.named_steps['lda'].scalings_,
    index=feature_names_style,
    columns=[f'LDA_Component_{i+1}' for i in range(lda_pipeline_style.named_steps['lda'].scalings_.shape[1])]
)
```

```python

# Gradient Boosting sobre o espaço LDA de Estilo
from sklearn.ensemble import GradientBoostingClassifier

X_train_style_lda = lda_pipeline_style.transform(X_train_style)
X_test_style_lda = lda_pipeline_style.transform(X_test_style)

gb_pipeline_style = Pipeline([
    ('scaler', StandardScaler()),
    ('gb', GradientBoostingClassifier(
        n_estimators=200,
        learning_rate=0.1,
        max_depth=3,
        random_state=42
    ))
])

print('Treinando Gradient Boosting no espaço LDA para Estilo...')
gb_pipeline_style.fit(X_train_style_lda, y_train_style)
print('Gradient Boosting treinado com sucesso!')

y_pred_style_gb = gb_pipeline_style.predict(X_test_style_lda)
print('\nRelatório de classificação - Gradient Boosting em Estilo:')
print(classification_report(y_test_style, y_pred_style_gb, zero_division=0))

cm_style_gb = confusion_matrix(y_test_style, y_pred_style_gb, labels=gb_pipeline_style.named_steps['gb'].classes_)
fig, ax = plt.subplots(figsize=(10, 8))
ConfusionMatrixDisplay(confusion_matrix=cm_style_gb, display_labels=gb_pipeline_style.named_steps['gb'].classes_).plot(ax=ax, cmap='Blues', colorbar=False)
ax.set_title('Confusion Matrix - Gradient Boosting LDA Estilo')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()

importances_style = gb_pipeline_style.named_steps['gb'].feature_importances_
component_names_style = [f'LDA_Component_{i+1}' for i in range(X_train_style_lda.shape[1])]
fig, ax = plt.subplots(figsize=(8, 4))
ax.bar(component_names_style, importances_style, color='orange')
ax.set_title('Gradient Boosting Importances - LDA Components Estilo')
ax.set_ylabel('Importance')
ax.set_xlabel('LDA Components')
ax.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```

# Recall @ K

```python
from sklearn.metrics.pairwise import cosine_similarity

# Recall@K for recommendation-style retrieval using LDA embeddings
# Exclui o próprio item da lista de candidatos.
X_retrieval = X_lda_style
labels_retrieval = y_style.reset_index(drop=True)

if X_retrieval.shape[0] != len(labels_retrieval):
    raise ValueError('Tamanho de X_retrieval diferente do número de rótulos.')

sim_matrix = cosine_similarity(X_retrieval)
np.fill_diagonal(sim_matrix, -np.inf)

k_values = [1, 5, 10, 20]
recall_at_k = {}

for k in k_values:
    hits = 0
    for idx in range(sim_matrix.shape[0]):
        top_k = np.argsort(-sim_matrix[idx])[:k]
        if labels_retrieval.iloc[idx] in labels_retrieval.iloc[top_k].values:
            hits += 1
    recall_at_k[k] = hits / sim_matrix.shape[0]

print('Recall@K usando LDA para Estilo:')
for k, score in recall_at_k.items():
    print(f'  Recall@{k}: {score:.4f}')

# Most frequent styles and their recall@5 as a quick recommendation quality check
style_scores = {}
for style in labels_retrieval.unique():
    idxs = labels_retrieval[labels_retrieval == style].index
    if len(idxs) < 2:
        continue
    hits = 0
    for idx in idxs:
        top_k = np.argsort(-sim_matrix[idx])[:5]
        if style in labels_retrieval.iloc[top_k].values:
            hits += 1
    style_scores[style] = hits / len(idxs)

style_scores = dict(sorted(style_scores.items(), key=lambda item: item[1], reverse=True))
```
