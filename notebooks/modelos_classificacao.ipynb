```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

PATH_DF_RECOMMENDED = '../data/processed/df_recommender.csv'

pd.set_option('display.max_columns',None)
pd.set_option('display.max_rows',None)
sns.set_style('darkgrid')
```

```python
df = pd.read_csv(PATH_DF_RECOMMENDED,  encoding='utf-8')
```

```python
display(df.isnull().sum())
display(df.columns)
display(df.info())
display(df.head())
```

```python
drops = ['Unnamed: 0']
df_class = df.copy()
df_class = df_class.drop(columns=drops)
```

```python
# preparando o dataset para classificacao
features_audio = [
    'tempo_bpm',
    'rms_mean', 'rms_std',
    'zcr_mean', 'zcr_std',
    'spectral_centroid_mean', 'spectral_centroid_std',
    'spectral_bandwidth_mean', 'spectral_bandwidth_std',
    'spectral_rolloff_mean', 'spectral_rolloff_std',
    'spectral_flatness_mean', 'spectral_flatness_std',
    'mfcc_01_mean', 'mfcc_02_mean', 'mfcc_03_mean', 'mfcc_04_mean',
    'mfcc_05_mean', 'mfcc_06_mean', 'mfcc_07_mean', 'mfcc_08_mean',
    'mfcc_09_mean', 'mfcc_10_mean', 'mfcc_11_mean', 'mfcc_12_mean',
    'mfcc_13_mean',
    'chroma_01_mean', 'chroma_02_mean', 'chroma_03_mean', 'chroma_04_mean',
    'chroma_05_mean', 'chroma_06_mean', 'chroma_07_mean', 'chroma_08_mean',
    'chroma_09_mean', 'chroma_10_mean', 'chroma_11_mean', 'chroma_12_mean',
    'onset_strength_mean', 'onset_strength_std',
]

# target (o que vamos prever)
target = 'ESTILO_MUSICAL'

# df_class = só features de áudio + target. Sem identificadores, sem metadados de vídeo, sem duplicatas.
df_class = df[features_audio + [target]].copy()

# sanity checks
print('Shape df_class:', df_class.shape)
print('Nulos em features:', df_class[features_audio].isnull().sum().sum())
print('Nulos em target:', df_class[target].isnull().sum())
print('Classes únicas:', df_class[target].nunique())
print()
print('Distribuição das classes:')
print(df_class[target].value_counts())
```

```python
is_null_target = df[df[target].isnull()].index
if len(is_null_target) > 0:
    df_class = df_class.drop(index=is_null_target)
    print('Valores nulos no taget removido')
else:
    print('Taget nao contem valores nulos')
```

```python
from sklearn.model_selection import train_test_split

X = df_class[features_audio]
y = df_class[target]

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,        
    stratify=y,            # mantém proporção das classes no train e test (CRÍTICO: nosso dataset é desbalanceado)
    random_state=42,       
)

print('Shape X_train:', X_train.shape)
print('Shape X_test :', X_test.shape)
print('Shape y_train:', y_train.shape)
print('Shape y_test :', y_test.shape)
print()
print('Distribuição no TREINO (% por classe):')
print((y_train.value_counts(normalize=True) * 100).round(2).head(10))
print()
print('Distribuição no TESTE (% por classe):')
print((y_test.value_counts(normalize=True) * 100).round(2).head(10))
```

### KNN

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import GridSearchCV

pipe_knn = Pipeline(steps=[
    ('scaler', StandardScaler()),
    ('knn', KNeighborsClassifier()),
])
```

```python
# grid
param_grid = {
    'knn__n_neighbors': [3, 5, 7, 11, 15, 21, 31],
    'knn__weights':     ['uniform', 'distance'],
}

grid_knn = GridSearchCV(
    estimator=pipe_knn,
    param_grid=param_grid,
    scoring='f1_macro',     # macro porque dataset é desbalanceado
    cv=5,                   
    n_jobs=-1,              # usa todos os núcleos da CPU
    return_train_score=True,
)

grid_knn.fit(X_train, y_train)

print('>> Melhor combinação de hiperparâmetros:')
print(grid_knn.best_params_)
print()
print(f'>> Melhor F1 macro (média no CV): {grid_knn.best_score_:.4f}')
print()
print('>> Top 5 combinações do grid:')
cv_results = pd.DataFrame(grid_knn.cv_results_)
print(
    cv_results[['param_knn__n_neighbors', 'param_knn__weights',
                'mean_test_score', 'std_test_score']]
    .sort_values('mean_test_score', ascending=False)
    .head(5)
    .to_string(index=False)
)
```

```python
import matplotlib.pyplot as plt

# pivot: linhas = n_neighbors, colunas = weights, valores = F1 médio
plot_df = cv_results.pivot_table(
    index='param_knn__n_neighbors',
    columns='param_knn__weights',
    values='mean_test_score',
)

ax = plot_df.plot(kind='bar', figsize=(8, 5), edgecolor='black')
ax.set_title('F1 macro médio (CV=5) por n_neighbors e weights')
ax.set_xlabel('n_neighbors (K)')
ax.set_ylabel('F1 macro médio')
ax.legend(title='weights')
ax.grid(axis='y', alpha=0.4)
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
from sklearn.metrics import (
    accuracy_score, f1_score, precision_score, recall_score,
    classification_report
)

# o GridSearchCV já refita automaticamente com os melhores hiperparâmetros em TODO o X_train
best_knn = grid_knn.best_estimator_

# predições no teste (que ainda não foi visto)
y_pred = best_knn.predict(X_test)

# métricas
print('Melhores hiperparâmetros:', grid_knn.best_params_)
print()
print(f'Accuracy        : {accuracy_score(y_test, y_pred):.4f}')
print(f'F1 macro        : {f1_score(y_test, y_pred, average="macro"):.4f}')
print(f'F1 weighted     : {f1_score(y_test, y_pred, average="weighted"):.4f}')
print(f'Precision macro : {precision_score(y_test, y_pred, average="macro", zero_division=0):.4f}')
print(f'Recall macro    : {recall_score(y_test, y_pred, average="macro", zero_division=0):.4f}')
print()
print('=== Classification Report (detalhado por classe) ===')
print(classification_report(y_test, y_pred, zero_division=0))
```

```python
from sklearn.metrics import confusion_matrix

def avaliar_modelo(nome_modelo, y_true, y_pred):
    """Retorna um dict com as métricas principais do modelo."""
    return {
        'modelo':            nome_modelo,
        'accuracy':          accuracy_score(y_true, y_pred),
        'precision_macro':   precision_score(y_true, y_pred, average='macro', zero_division=0),
        'recall_macro':      recall_score(y_true, y_pred, average='macro', zero_division=0),
        'f1_macro':          f1_score(y_true, y_pred, average='macro', zero_division=0),
        'f1_weighted':       f1_score(y_true, y_pred, average='weighted', zero_division=0),
    }

# inicializa o df_metrics (só na primeira vez — depois é só dar append)
df_metrics = pd.DataFrame(columns=[
    'modelo', 'accuracy', 'precision_macro', 'recall_macro', 'f1_macro', 'f1_weighted'
])

# registra o KNN
metricas_knn = avaliar_modelo('KNN', y_test, y_pred)
df_metrics = pd.concat([df_metrics, pd.DataFrame([metricas_knn])], ignore_index=True)
df_metrics.head()


cm = confusion_matrix(y_test,y_pred)

plt.figure(figsize=(10,5))
sns.heatmap(
    cm,
    annot=True,
    fmt='d',
    cmap='Blues',
    xticklabels= best_knn.classes_,
    yticklabels= best_knn.classes_
)
plt.title('Matrix de Confusao')
plt.xlabel('Previsto pelo Modelo')
plt.ylabel('Real')
plt.show()


```

## Conclusao

Avaliação do modelo KNN para classificação de estilo musical
Contexto do dataset. O dataset contém 3.890 amostras distribuídas em 35 classes de estilo musical, com forte desbalanceamento (razão entre classe majoritária e minoritária de 314:1 — Sertanejo: 629, Piseiro: 2). Treze classes possuem menos de 15 amostras totais, o que limita estatisticamente qualquer modelo de aprender padrões para elas. Quatro amostras com target nulo foram removidas.

Justificativa da métrica. Adotou-se f1_macro como métrica de otimização no GridSearch porque o objetivo de negócio (recomendar bandas de qualquer estilo) exige tratar todas as classes com peso igual. Métricas ponderadas por suporte (f1_weighted, accuracy) recompensariam um modelo que acerta apenas as classes majoritárias — exatamente o comportamento que se quer detectar e evitar.

Resultado do GridSearch. A melhor combinação encontrada (n_neighbors=5, weights='distance') atingiu f1_macro = 0.0876 ± 0.019 na validação cruzada. Crucialmente, as 5 melhores combinações testadas variam entre 0.077 e 0.088, ou seja, não existe um hiperparâmetro neste espaço de busca que faça o modelo aprender de forma satisfatória. Isso indica que o problema é estrutural (modelo ou features), não de tuning.

Avaliação no conjunto de teste. Os resultados foram:

1. Accuracy: 0.208 — apenas 4,6 pontos percentuais acima do baseline majoritário (0.162), que sempre prediria Sertanejo.
2. F1 macro: 0.070 — embora 8,75× superior ao baseline (0.008), o valor absoluto é incompatível com qualquer uso prático.
3. F1 weighted: 0.184 — discrepância de 2,6× em relação ao f1_macro confirma o viés do modelo para classes majoritárias.
4. Precision macro (0.078) ≈ Recall macro (0.074) — modelo é uniformemente ruim, não há ajuste de threshold ou viés de classificação a explorar.
5. O resultado do teste (0.070) está dentro de um desvio-padrão do CV (0.088 ± 0.019), descartando overfitting. A limitação é capacidade preditiva, não generalização.
6. Análise por classe (classification_report). 21 das 33 classes presentes no teste possuem F1 = 0,00, ou seja, o modelo nunca classifica corretamente essas categorias. As únicas classes com desempenho relevante são as duas maiores do dataset (Sertanejo: f1=0.40 com recall=0.56; MPB: f1=0.29). 
7. Classes médias (Pop Rock, Pagode, Samba, Rock) ficam entre f1=0.14 e 0.19. O F1=0.40 da classe Axé é estatisticamente irrelevante (1 acerto em 2 amostras de teste).

O modelo KNN, mesmo após padronização (StandardScaler) e busca de hiperparâmetros via GridSearch com 5-fold CV, não é adequado para esta tarefa. As evidências são:

    Performance absoluta inaceitável em todas as métricas que tratam classes com peso igual.
    Nenhuma combinação de hiperparâmetros testada produziu f1_macro > 0,09.
    O modelo só apresenta sinal em 2 das 35 classes — exatamente as majoritárias —, o que viola o objetivo do projeto.

**Recomendação.** Descartar o KNN como modelo final, mas registrar três limitações estruturais que afetariam qualquer modelo neste dataset e devem ser endereçadas antes de testar alternativas:

1. Granularidade excessiva do target: classes como "Pop", "Pop Rock", "Rock", "Indie/Alternativo" e provavelmente são indistinguíveis no espaço de features de áudio. Sugere-se reagrupar para 8-12 macro-classes. Sugestão é agrupar os estilos de musicas que são raros em um grupo chamada "Outros".
2. KNN não é apropriado para alta dimensionalidade: com 40 features, distâncias euclidianas perdem poder discriminativo. Para próximas iterações, testar Random Forest ou Gradient Boosting, que toleram melhor essas condições.
